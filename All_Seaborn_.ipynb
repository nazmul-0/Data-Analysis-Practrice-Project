```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Loading The dataset

```python
mart = pd.read_excel('/content/supermarket_sales.xlsx')
mart.head()
```

# Countplot

countplot with values

```python
ax = sns.countplot(mart,x='Gender')
ax.bar_label(ax.containers[0])
```

Title of the bars weren't visible so , increased the figsize

```python
plt.figure(figsize=(15,5))
ax=sns.countplot(mart, x= 'Product line')
ax.bar_label(ax.containers[0])
```

Showing countplot horizontally

```python
plt.figure(figsize=(15,5))
ax=sns.countplot(mart, y= 'Product line')
ax.bar_label(ax.containers[0])
```

showing color plot with hue

```python
plt.figure(figsize=(12,8))
ax=sns.countplot(mart, x= 'Product line', hue = 'Gender')
ax.bar_label(ax.containers[0])
```

adding a color pallette

```python
plt.figure(figsize=(12,8))
ax=sns.countplot(mart, x= 'Product line', hue = 'Gender',palette = 'muted')
ax.bar_label(ax.containers[0])
```

```python
plt.figure(figsize=(15,5))
sns.countplot(mart, x= 'Product line', facecolor = (0,0,0,0),linewidth = 5, edgecolor = sns.color_palette('dark',3))

```
